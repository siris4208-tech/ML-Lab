```python
import pandas as pd
df=pd.read_csv("tested.csv")
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))

plt.hist(df["Age"].dropna(),
         bins=15,
         color="skyblue",
         edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```
